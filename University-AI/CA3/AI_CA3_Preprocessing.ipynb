## Importing Require Libraries

```python
import pandas as pd
from sklearn.impute import SimpleImputer
```

## Reading The Input Database

```python
def read_csv_file(file_path):
    return pd.read_csv(file_path)
```

## Handling Missing Values

```python
def impute_missing_values(df):
    num_cols = df.select_dtypes(include='number').columns
    cat_cols = df.select_dtypes(include='object').columns
    num_imputer = SimpleImputer(strategy='mean')
    cat_imputer = SimpleImputer(strategy='most_frequent')
    df[num_cols] = num_imputer.fit_transform(df[num_cols])
    df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
    return df
```

## Droping Unnecessary Columns

```python
def drop_unnecessary_columns(df):
    columns_to_drop = []
    if 'university' in df.columns:
        columns_to_drop.append('university')
    if 'fatherJob' in df.columns:
        columns_to_drop.append('fatherJob')
    if 'motherJob' in df.columns:
        columns_to_drop.append('motherJob')
    df = df.drop(columns=columns_to_drop)
    return df
```

## Incoding Binary Categorical Columns

```python
def encode_binary_columns(df):
    binary_map = {'yes': 1, 'no': 0, 'F': 0, 'M': 1, 'U': 1, 'R': 0}
    binary_cols = ['sex', 'address', 'universitySupport', 'paid', 'higher', 'internet', 'romantic']
    for col in binary_cols:
        if col in df.columns:
            df[col] = df[col].map(binary_map)
    return df
```

## Incoding Multi-Class Categorical Columns

```python
def encode_multicategory_columns(df):
    multi_cat_cols = ['reason']
    df = pd.get_dummies(df, columns=[c for c in multi_cat_cols if c in df.columns], drop_first=True)
    df = df.astype(int)
    return df
```

## New Step: Converting Final Grade to Categories

Since we plan to use classification models in the next steps, we need to transform the continuous finalGrade values into categorical labels. Classification models are designed to work with discrete classes, not continuous numerical values.

We divided the final grades into four categories:

- A: grade ≥ 17  
- B: 14 ≤ grade < 17  
- C: 10 ≤ grade < 14  
- F: grade < 10  

A new column named grade_category was added to the dataset to store these class labels.

```python
def convert_grade_to_category(df):
    def categorize(grade):
        if grade >= 17:
            return 'A'
        elif grade >= 14:
            return 'B'
        elif grade >= 10:
            return 'C'
        else:
            return 'F'
    
    df['grade_category'] = df['finalGrade'].apply(categorize)
    return df
```

## Saving The Preprocessed Dataset

```python
def save_processed_data(df, output_path):
    df.to_csv(output_path, index=False)
```

## Preprocessing Pipeline Execution

```python
def preprocess_data(input_csv_path, output_csv_path):
    df = read_csv_file(input_csv_path)
    df = impute_missing_values(df)
    df = drop_unnecessary_columns(df)
    df = encode_binary_columns(df)
    df = encode_multicategory_columns(df)
    df = convert_grade_to_category(df)
    save_processed_data(df, output_csv_path)
    return df
```

## Running The Preprocession On The Dataset

```python
input_path = "Grades.csv"
output_path = "Grades_processed.csv"
df = preprocess_data(input_path, output_path)
df.head()
```
